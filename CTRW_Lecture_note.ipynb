# Monte Carlo (MC) Simulation

여러분이 아래에서 보실 시뮬레이션은 [Monte Carlo Simulation](https://en.wikipedia.org/wiki/Monte_Carlo_method) 방식입니다.

이 시뮬레이션 방식은 확률론적 기법을 사용해서 문제를 해결하는 방법으로, 주로 수학적, 물리적, 경제적 시스템에 사용됩니다.

Monte Carlo 시뮬레이션의 역사에 대해 간략히 알려드리겠습니다.:
1. 기원과 초기 발전

    1940년대 초반: Monte Carlo 시뮬레이션의 기원은 제2차 세계 대전 기간 동안 미국에서 맨해튼 프로젝트를 수행하던 과학자들에 의해 시작되었습니다. 이 프로젝트는 핵무기 개발을 위한 것으로, 당시 로스 알라모스 연구소에서 일하던 Stanislaw Ulam과 John von Neumann이 무작위 샘플링 기법을 통해 복잡한 물리적 문제를 해결할 방법을 모색했습니다.

    ![image](https://upload.wikimedia.org/wikipedia/commons/thumb/5/5e/JohnvonNeumann-LosAlamos.gif/220px-JohnvonNeumann-LosAlamos.gif)

    (천재중에 천재라고 불렸던 John von Neumann, 폰 노이만이라고 쓰면 알아보실까요?)

    폰 노이만이 이 시뮬레이션을 통해 진행한 연구는 다음과 같습니다.

    1. 핵무기 개발: 폰 노이만은 맨해튼 프로젝트에 참여하면서 핵분열 과정을 모델링하는 데 몬테카를로 방법을 사용했습니다. 이 시뮬레이션 기법은 핵 분열과 연쇄 반응의 확률을 계산하고, 폭발의 성능을 예측하는 데 중요했습니다.

    2. 중성자 수송 문제: 핵무기의 효율적인 설계를 위해 중성자의 확산과 흡수를 분석하는 문제가 중요했습니다. Monte Carlo 시뮬레이션은 이러한 확산 과정을 무작위적으로 샘플링하여 해결하는 데 사용되었습니다.

    3. 수소폭탄 개발: 폰노이만은 수소폭탄 개발에도 Monte Carlo 시뮬레이션을 사용했습니다. 그는 이 시뮬레이션을 통해 폭발의 열역학적 특성을 연구하고, 수소폭탄의 폭발력을 최적화하는 데 중요한 역할을 했습니다.

    4. 기체 역학 및 유체 역학: 폰 노이만은 Monte Carlo 시뮬레이션을 이용해서 기체 역학 및 유체 역학 문제를 풀었습니다. 특히 복잡한 유체 흐름과 관련된 문제를 해결하는데 기여를 했습니다.

2. 이름의 유래

    Monte Carlo 시뮬레이션이라는 이름은 Monte Carlo 카지노의 이름에서 유래했습니다. Ulam의 동료인 Nicholas Metropolis가 이 방법을 제안할 때, 무작위성과 게임의 우연성을 상징하는 Monte Carlo의 이름을 따서 명명하였습니다.

3. 컴퓨터의 발전과 Monte Carlo 시뮬레이션

    1940-1950년대: 초창기 컴퓨터의 등장과 함께 Monte Carlo 방법이 더 널리 사용되기 시작했습니다. 당시 ENIAC과 같은 초기 컴퓨터들은 이 방법을 적용하여 복잡한 계산을 수행할 수 있었습니다.

    1950년대 중반: Monte Carlo 시뮬레이션은 다양한 분야에서 응용되기 시작했습니다. 물리학, 공학, 금융, 그리고 기타 여러 분야에서 불확실성을 다루는 데 유용한 도구로 자리잡았습니다.

4. 현대적 발전

    1980년대 이후: 컴퓨터 성능의 비약적인 발전과 함께 Monte Carlo 시뮬레이션은 더욱 정교해졌습니다. 고성능 컴퓨팅과 병렬처리 기법의 도입으로 매우 복잡한 시스템의 시뮬레이션도 가능해졌습니다.

    금융 및 경제 분야: 1990년대 이후로는 금융 리스크 관리, 옵션 가격 결정, 투자 포트폴리오 분석 등 다양한 경제적 모델링에도 널리 사용되고 있습니다.
    기타 분야: 기후 모델링, 인공지능, 의학 연구 등 다양한 현대 과학 및 산업 분야에서도 Monte Carlo 시뮬레이션은 필수적인 도구로 자리잡고 있습니다.

Monte Carlo 시뮬레이션은 이렇게 과거부터 현재까지 다양한 발전을 거치며, 많은 분야에서 중요한 도구로 사용되고 있습니다.

## Simulation에 대한 전반적인 소개

1. Uniform distribution $f(U)$에서 random number 얻기

    Uniform distribution $f(U)$에서 변수 $U$는 [0,1] 사이의 실수값을 가지며, $f(U)$는 다음과 같은 식을 만족합니다.

    $$f(U) = 1 \text{  for   } 0 ≤ U ≤ 1$$

2. random number $U$에 해당되는 random time $t$를 찾기

    [Inverse Transform Sampling](https://m.blog.naver.com/jinis_stat/221648391742) 방법을 이용해 random number $U$에 해당되는 random time $t$를 찾습니다.
    
    (여러분들에게 나눠준 zip 파일에 ITM.pdf 파일을 참조해주세요.)

    제목은 거창하지만, 핵심만 요약하면 다음과 같습니다.
    
    random number $U$와 random time $t$, 그리고 이 random time이 따르는 waiting time distribution $\psi(t)$는 다음과 같은 조건을 만족하게 됩니다.

    $$U=\int_{0}^{t} \psi(\tau) d\tau = 𝔯(t) → t = 𝔯^{-1}(U)$$

    이 내용은 한번 ITM을 여러분들이 직접 증명해볼 것을 권장드리고 싶습니다.

3. random time $t$를 이용해 원하는 값들을 도출하기

    waiting time distribution, mean, variance 같은 값들이나, 이러한 time들의 convolute된 값들을 구하실 수 있게 됩니다.

python을 어느정도 아신다는 가정 하에 아래 code를 작성해봤습니다.

최대한 읽기 편하게 쓴다고 써봤는데.. code의 내용을 읽어보시고, 이해가 잘 안되는 부분이 있으시다면 질문을 해주시면 됩니다.

위의 그림에 나와있는 내용을 토대로, exponential distribution $\psi(t)= k e^{-kt}$에서 random time을 구하는 함수를 아래와 같이 구현해봤습니다.

$$U=1-e^{-kt}$$

$$\ln(1-U) = -kt$$

$$\therefore t = -\frac{\ln(1-U)}{k}$$

$U$와 $\ln$ 함수를 구현하기 위해, 'numpy'라는 파이썬 라이브러리를 사용했고,

이를 graph로 보이기 위해 'matplotlib.pyplot'이라는 라이브러리를 사용해봤습니다.

```python
import numpy as np
import matplotlib.pyplot as plt

def psi(k,t):
    return k*np.exp(-k*t)

t_array = np.linspace(0,10,1000)

def psi_waiting_time(k, nr):
    """
    Generates a random time from the waiting time distribution psi(t) = k e^-kt.

    Repeat nr times.

    Args:
        k: The rate parameter of the distribution.
        nr: The number of times to repeat the process.

    Returns:
        A random time array generated from the distribution.
    """
    return -np.log(1-np.random.uniform(0, 1,nr)) / k
```

위에 정의한 함수는 총 2가지입니다.

psi(k,t) 함수는 exponential distribution을 그리기 위해 정의한 함수입니다.

psi_waiting_time(k,nr) 함수는 nr개의 random time을 generating하는 함수입니다.

그러면, 이 함수가 얼마나 잘 먹히는지 한번 알아봅시다.

```python
array = psi_waiting_time(1, 100) # k = 1, 100번 반복
plt.hist(array, 1000, density = True, label = 'simulation result')
plt.plot(t_array, psi(1,t_array), lw = 4, label = "$k e^{-kt}$")
plt.legend(loc = 'upper right')
plt.show()
```

```python
array = psi_waiting_time(1, 1000) # k = 1, 1000번 반복
plt.hist(array, 1000, density = True, label = 'simulation result')
plt.plot(t_array, psi(1,t_array), lw = 4, label = "$k e^{-kt}$")
plt.legend(loc = 'upper right')
plt.show()
```

```python
array = psi_waiting_time(1, 100000) # k = 1, 100000번 반복
plt.hist(array, 1000, density = True, label = 'simulation result')
plt.plot(t_array, psi(1,t_array), lw = 4, label = "$k e^{-kt}$")
plt.legend(loc = 'upper right')
plt.show()
```

위의 결과를 보면 알 수 있듯이, 반복 횟수가 증가할수록, waiting time을 이용해 만든 histogram이 exponential distribution을 따라간다.

### exponential distribution에서 random time을 반복해서 뽑았을 때, 이 값들의 평균과 분산을 구해보고, 이를 이론과 비교해보자.

```python
nr = [] # 반복하는 횟수를 기록할 빈 array
mean = [] # 평균값을 기록할 빈 array
variance = [] # 분산을 기록할 빈 array
k = 1

for i in range(0,25):
    number = 2**i # 반복하는 횟수는 2, 4, 8, 16, ... 으로 증가한다.
    array = psi_waiting_time(k, number) # 여기에서 random time을 뽑는다.

    mean_time = np.mean(array) # 평균을 계산한다.
    variance_time = np.mean(array**2) - mean_time**2 # 제곱의 평균 - 평균의 제곱, 분산을 계산한다. 여기에서 a**b는 "a의 b승" 을 의미합니다.

    nr.append(number) # 결과 값을 array에 넣어준다.
    mean.append(mean_time) # 결과 값을 array에 넣어준다.
    variance.append(variance_time) # 결과 값을 array에 넣어준다.
```

```python
# 반복 횟수에 따른 mean time의 변화

plt.plot(nr, mean, label = 'simulation')
plt.plot([1,2**25],[1/k,1/k], label = 'expected value', lw = 4)
plt.xscale('log')
plt.xlabel('number of repetation')
plt.ylabel('mean time')
plt.legend(loc = 'upper right')
```

```python
# 반복 횟수에 따른 variance of time의 변화

plt.plot(nr, variance, label = "simulation")
plt.plot([1,2**25],[1/k,1/k], label = "expected value", lw = 4)
plt.xscale('log')
plt.xlabel('number of repetation')
plt.ylabel('variance of time')
plt.legend(loc = 'upper right')
```

위의 결과를 보면 알 수 있듯이, 평균과 분산을 구할 때 random time을 많이 뽑을수록, 저희가 생각하는 이론의 값에 수렴한다는 사실을 확인하실 수 있습니다.

# 경쟁 process에서 나오는 waiting time distribution

그렇다면, 경쟁 process는 어떻게 표현하면 좋을까요?

image.png

위의 상황을 시뮬레이션 하는 방법에 대해 생각해봅시다. 왼쪽으로 움직일지, 오른쪽으로 움직일지 결정하는 것은 "어떤 event가 더 먼저 발생했느냐" 입니다.

이를 시뮬레이션으로 표현해보겠습니다.

```python
def psi_left_win(kL,kR,nr):
    left = psi_waiting_time(kL, nr)
    right = psi_waiting_time(kR, nr)

    return [time for time in left * (left < right) if time !=0]

def psi_right_win(kL,kR,nr):
    left = psi_waiting_time(kL, nr)
    right = psi_waiting_time(kR, nr)

    return [time for time in right * (left > right) if time !=0]

def psi_win(kL,kR,nr):
    left = psi_waiting_time(kL, nr)
    right = psi_waiting_time(kR, nr)

    return right * (left > right) + left * (left < right)

def probability_of_win_left(kL,kR,nr):
    left = psi_waiting_time(kL, nr)
    right = psi_waiting_time(kR, nr)

    return np.sum(left < right) / nr

def probability_of_win_right(kL,kR,nr):
    left = psi_waiting_time(kL, nr)
    right = psi_waiting_time(kR, nr)

    return np.sum(left > right) / nr
```

위에서 (left > right), (left < right) 라고 표현한 것은 Boolean Array입니다.

이는 다음과 같이 작동합니다.


>    e.g. left = [1,2,3,4], right = [4,3,2,1]
>
>    (left > right) = [0,0,1,1], (left < right) = [1,1,0,0]


각각의 array의 값을 확인하고, 제가 준 조건이 맞으면 1을, 틀리면 0의 값을 뱉어냅니다.

"왜 이런걸 쓰나요?" 라는 질문을 하실 수 있는데,

요즘 공부하고 계시는 C 언어에서는 이 과정을 for loop로 해결해도 문제가 되지 않지만,

python, matlab에서는 여러개의 random number를 한번에 뽑고, Boolean Array를 이용하는 것이 더 빠르게 작동하기 때문에 사용합니다.

이런건 테크닉적인 부분이라, 당장 무시하셔도 됩니다.

아래의 결과를 한번 봐보죠.

```python
array = psi_win(0.5, 0.5, 100000)
plt.hist(array, 1000, density = True)
plt.plot(t_array, psi(1,t_array), lw = 4)
plt.show()
```

```python
array = psi_win(0.2, 0.8, 100000)
plt.hist(array, 1000, density = True)
plt.plot(t_array, psi(1,t_array), lw = 4)
plt.show()
```

위의 결과를 보면 다음과 같은 사실을 알 수 있습니다.

> 왼쪽으로 뛰는 rate와 오른쪽으로 뛰는 rate가 (0.5, 0.5)인 경우와 (0.2, 0.8)인 경우에, 경쟁 process를 통해 나온 waiting time의 distribution은 같다.

놀랍지 않나요? 후자의 경우, 오른쪽으로 뛰는 rate는 왼쪽으로 뛰는 rate에 비해 4배 크니, 거의 대부분 오른쪽으로 뛸탠데..

그럼에도 불구하고, 왼쪽으로 뛰는 경우가 발생하고 있으며, 이 경쟁으로 인해 (0.5, 0.5)의 rate를 가지는 경우와 waiting time distribution이 같게 됩니다.

```python
probability_of_win_right(0.5,0.5,10000000)
```

```python
probability_of_win_left(0.5,0.5,10000000)
```

```python
probability_of_win_right(0.2,0.8,10000000)
```

```python
probability_of_win_left(0.2,0.8,10000000)
```

위에 'probability_of_win_right'와 'probability_of_win_left' 함수는 '왼쪽이 이기는 경우의 수 / 총 반복 횟수' 를 계산해주는 함수입니다. 보시다싶이, 저희가 예측한 대로의 값이 나옵니다.

## Random walk simulation 구현

```python
def random_walker(kL,kR,m,t):
    """
    Args:
    kL, kR: rate coefficient
    m: initial position
    t: measuring time (final)

    Returns:
        position of random walker which starts at lattice site m.

    """
    time = 0
    position = m

    while time < t:
        tL = psi_waiting_time(kL, 1)[0]
        tR = psi_waiting_time(kR, 1)[0]

        if tL < tR:
            position -= 1
            time += tL
        elif tR < tL:
            position += 1
            time += tR

    return position

def normal_distribution(sigma,x,m):
    return np.exp(-0.5*((x-m)/sigma)**2) / np.sqrt(2*np.pi * sigma**2)

x_axis = np.linspace(-150,150,301)
```

```python
nr = 4000
time = 1000

results = np.zeros(nr)

for i in range(0,nr):
    results[i] = random_walker(1,1,0,time)
```

```python
plt.hist(results, 200, density = True)
plt.plot(x_axis, normal_distribution(50,x_axis,0), lw = 4)
plt.plot()
plt.show()
```

4000번 반복해서 1000의 시간까지 시뮬레이션을 돌렸을 때, Gaussian에 근접하게 나온 것 같은데... 하지만 완벽하게 일치하는 결과를 얻지는 못했습니다.

Boolean Array를 사용한 코드는 위의 코드와 비교했을 때 더 빠르게 계산이 됩니다.

제 컴퓨터 기준으로 같은 시간동안에 100000번 반복해서 10000의 시간까지 계산하는 것이 가능해집니다.

(반복횟수는 25배, 시간은 10배 큰 시뮬레이션입니다.)



```python
def random_walker_parallel(kL,kR,m,t,nr):
    """
    Args:
    kL, kR: rate coefficient
    m: initial position
    t: measuring time (final)
    nr: number of repeatation.

    Returns:
        positions of random walkers which starts at lattice site m.

    """
    time = np.zeros(nr)
    position = m*np.ones(nr)

    while True:
        tL = psi_waiting_time(kL, nr)
        tR = psi_waiting_time(kR, nr)

        B_left = (tL < tR)
        B_right = (tL > tR)

        time += tL * B_left + tR * B_right
        B_time = (time < t)


        position -= B_left * B_time
        position += B_right * B_time

        if np.sum(B_time) == 0:
            break

    return position
```

```python
nr = 100000
time = 10000

results = random_walker_parallel(1,1,0,time, nr)
```

```python
x_axis = np.linspace(-600,600,1201)

plt.hist(results, 200, density = True)
plt.plot(x_axis, normal_distribution(130,x_axis,0), lw = 4)
plt.plot()
plt.show()
```

### exponential distribution에서 얻은 random time을 n번 더한 시간을 nr번 뽑았을 때, 이 시간이 만드는 distribution은 어떻게 되는가?

```python
def psi_waiting_time_sum(k, nr,n):
    """
    Generates a random time from the waiting time distribution psi(t) = k e^-kt.

    Repeat nr times.

    Args:
        k: The rate parameter of the distribution.
        nr: The number of times to repeat the process.
        n: The number of times to summation
    Returns:
        A random time array generated from the distribution.
    """
    return np.sum(-np.log(np.random.uniform(0, 1,(nr,n))) / k, axis = 1)

from scipy.special import gamma

def Gamma_distribution(n,k,t):
    return np.exp(-k*t)* k**n * t**(-1+n) / gamma(n)
```

```python
array = psi_waiting_time_sum(1, 100000, 3)
```

```python
plt.hist(array, 1000, density = True)
plt.plot(t_array, Gamma_distribution(3,1,t_array), lw = 4)
plt.show()
```

이론적으로 생각해보면, $\psi(t)$에서 뽑은 random number를 n번 더한 시간은 다음과 같은 distribution (우선 $\phi(t)$라고 지칭해봅시다.)을 따릅니다.

$$\hat{\phi}(u) = \left(\hat{\psi}(u)\right)^n$$

이론적으로는 이것이 맞는데, 직접 simulation으로 입증해보는 것은 다른 느낌으로 다가올 것이라고 생각이 됩니다.
