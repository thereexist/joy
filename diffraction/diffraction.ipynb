# 프리즘과 회절격자에서 나타나는 분산 비교

빛의 분산은 일반인들에게는 무지개 또는 프리즘에서 입사된 백색광이 빨주노초파남보의 색으로 나뉘는 것으로 알려진, 꽤 상식으로 여겨지는 현상이다. 

그러나 아마 일반인들은 현대 분광학 연구에서 빛을 파장별로 나누는데 프리즘을 사용하지 않는다는 사실은 모를 것이다.

현대 광학에서 빛을 파장별로 나누는데 가장 널리 사용되는 방법은 grating(회절격자)를 활용하는 것이다.

"왜 프리즘을 사용하지 않는가?" 라는 질문에 대해 한번 설명을 해보려고 한다.

아마도 글을 생각없이 막 써내려가서, 친절하게 받아들여지지는 않을 것 같다.

나중에 수정을 해보겠다.

# 1. Prism에서의 분산

위의 그림은 각 변이 60도인 정삼각형 프리즘에 $\theta_1$의 각도로 입사된 빛이 Snell's law에 의해 굴절되는 상황을 보여주고 있다.

최종적으로 프리즘에서 나가는 빛이 프리즘의 수직한 면과 이루는 각도는 $\theta_3$로, 이는 복잡하지는 않지만, 꽤 더러운 식을 풀어야 풀리는 결과이다.

이 결과를 통해 처음 입사된 빛에 비해 틀어진 각도를 계산해보면 $\theta_3+\theta_1-60\degree$ 가 된다는 것을 확인할 수 있다.

이때 $\theta_3$의 각을 계산하는데 refractive index가 주요한 factor로 작용한다. 이 refractive index는 입사하는 빛의 파장 $\lambda$에 대한 식으로 나타나며 이 관계는 선형적이지 않다.

이 refractive index와 빛의 파장의 관계는 [link](*https://refractiveindex.info/?shelf=glass&book=fused_silica&page=Malitson*) 와 같은 사이트에서 확인이 가능하다. 물질별로 다르게 나온다는 것을 확인 가능하다.

나는 여기에서, Fused silica 물질에 대해 진행을 해보려고 한다. 이 물질에서 $\lambda$와 refractive index의 관계식은 다음과 같다.

(Malitson 1965: n 0.21-3.71 μm 기준)

$$n^2-1=\frac{0.6961663\lambda^2}{\lambda^2-0.0684043^2} + \frac{0.4079426\lambda^2}{\lambda^2-0.1162414^2} + \frac{0.8974794\lambda^2}{\lambda^2-9.896161^2} $$




위의 식을 통해 알 수 있듯이, 프리즘의 경우 빛의 경로가 틀어진 정도를 계산하는데, 어렵지는 않지만 지저분한 식이 많이 들어간다.

이 식을 통해 얻은 결과는 어떨까?

아마 (이미 답을 아는 상태에서 아마라고 하는 것도 웃기지만) '파장의 변함'에 따른 '빛의 경로가 틀어진 경로'는 선형적이지 않을 것이며, 왜곡될 것이다.

이는 특정 파장의 빛만 쪼개서 얻거나, 빛을 분산해서 분석하는데 매우 안좋은 조건이다.


# 2. Grating에서의 분산

우선 graiting (회절격자)에 대해 낯선 분들이 있다면, 집에 있는 CD나 DVD (요즘은 많이 안쓰는 물건이기도 하다.)에 백색광을 비춰보자. 

그렇게 하면 빛이 빨주노초파남보의 색으로 갈라지는 것을 눈으로 확인할 수 있을 것이다.


물리학과에서 광학을 배우게 되면, 일정한 간격으로 격자를 놓는 것 만으로도 빛의 파장을 가를 수 있다는 재미있는(?) 공부를 할 수 있다.

이것이 grating인데, CD나 DVD는 reflection grating의 일종으로 볼 수 있고, 앞으로 다뤄볼 grating은 Transmission grating이다.

이 Transmission grating에서 빛의 경로가 틀어진 정도는 다음과 같이 계산이 된다.

$m\lambda = d(\sin{\alpha} - \sin{\beta})$의 식을 보면 겁을 먹겠지만, 그렇게 어렵지 않다.

고등학교에서 배운 영의 이중 슬릿 실험 (Double-slit experiment)를 기억하실련지?

정확히 같은 원리라고 볼 수 있다.

Grating(회절격자)에서 투과되는 빛이 매우 많이 있을 탠데 (영의 이중슬릿에서는 2개만 있었는데!), 

이 빛들이 보강 간섭이 일어나는 경로를 계산하면 $m\lambda = d(\sin{\alpha} - \sin{\beta})$ 식이 나온다. 

이 식을 이용하면 빛의 경로가 틀어진 정도는 grating을 구성하는 물질에 상관 없이 항상 같게 표현 가능하다. 

(여기에서 $\alpha$나 $\beta$의 값이 충분히 작다면, radian 각 기준으로, $\Delta\theta ≈ \lambda/d$ 가 성립한다. 이것이 장점인데, radian 각에 익숙하지 않은 사람이 있을까봐, 도수법 (360도로 표현하는 방법)을 사용했다.)

그렇다면, 가시광선의 영역 (대략 380 - 750 nm)에서 빛을 prism으로 분산시킬때와, grating으로 분산시킬때, 어떻게 달라지는지를 한번 시각화해서 봐보자.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# ======= 1. Sellmeier equation for fused silica =======
def sellmeier_fused_silica(wavelength_nm):
    wavelength_um = wavelength_nm / 1000
    B1 = 0.6961663
    B2 = 0.4079426
    B3 = 0.8974794
    C1 = 0.0684043**2
    C2 = 0.1162414**2
    C3 = 9.896161**2
    lambda_sq = wavelength_um ** 2
    n_squared = 1 + (B1 * lambda_sq) / (lambda_sq - C1) \
                  + (B2 * lambda_sq) / (lambda_sq - C2) \
                  + (B3 * lambda_sq) / (lambda_sq - C3)
    return np.sqrt(n_squared)

# ======= 2. Prism deviation angle calculation =======
def prism_deviation_angle(n1, n2, theta1_deg):
    theta1_rad = np.radians(theta1_deg)
    sin_theta2 = (n1 / n2) * np.sin(theta1_rad)
    theta2_rad = np.arcsin(sin_theta2)
    internal_angle = np.radians(60) - theta2_rad
    sin_theta3 = (n2 / n1) * np.sin(internal_angle)
    theta3_rad = np.arcsin(sin_theta3)
    deviation_deg = np.degrees(theta3_rad + theta1_rad - np.radians(60))
    return deviation_deg

# ======= 3. Diffraction grating deviation angle =======
def diffraction_deviation(alpha_deg, m, wavelength_nm=500, grating_spacing_nm=1000):
    alpha_rad = np.radians(alpha_deg)
    lambda_over_d = (m * wavelength_nm) / grating_spacing_nm
    sin_beta = np.sin(alpha_rad) - lambda_over_d
    if np.abs(sin_beta) > 1:
        raise ValueError("No real solution for beta.")
    beta_rad = np.arcsin(sin_beta)
    delta_theta_deg = alpha_deg - np.degrees(beta_rad)
    return delta_theta_deg
```

위의 함수 3개를 통해, 파장에 따른 refractive index, prism을 통과한 빛과 입사한 빛 사이의 각도 차이, grating을 통과한 빛과 입사한 빛 사이의 각도 차이를 구하는 함수를 구현했다.

이제 이 함수를 활용해서 시각화를 해보자.

```python
# ======= 4. RGB Color Mapping =======
def wavelength_to_rgb(wavelength):
    gamma = 0.8
    factor = 1.0
    R = G = B = 0.0
    if 380 <= wavelength <= 440:
        R = -(wavelength - 440) / (440 - 380)
        G = 0.0
        B = 1.0
    elif 440 < wavelength <= 490:
        R = 0.0
        G = (wavelength - 440) / (490 - 440)
        B = 1.0
    elif 490 < wavelength <= 510:
        R = 0.0
        G = 1.0
        B = -(wavelength - 510) / (510 - 490)
    elif 510 < wavelength <= 580:
        R = (wavelength - 510) / (580 - 510)
        G = 1.0
        B = 0.0
    elif 580 < wavelength <= 645:
        R = 1.0
        G = -(wavelength - 645) / (645 - 580)
        B = 0.0
    elif 645 < wavelength <= 750:
        R = 1.0
        G = 0.0
        B = 0.0
    R = (R * factor) ** gamma if R > 0 else 0
    G = (G * factor) ** gamma if G > 0 else 0
    B = (B * factor) ** gamma if B > 0 else 0
    return (R, G, B)

# ======= 5. Main simulation and plotting =======
wavelengths_nm = np.linspace(380, 750, 500)
colors = [wavelength_to_rgb(wl) for wl in wavelengths_nm]

# Prism
n1 = 1.0
theta1_deg = 50
n2_list = sellmeier_fused_silica(wavelengths_nm)
deviation_prism = [prism_deviation_angle(n1, n2, theta1_deg) for n2 in n2_list]

# Grating
m = 1
grating_spacing_nm = 1000
alpha_deg = 30
deviation_grating = []
for wl in wavelengths_nm:
    try:
        delta = diffraction_deviation(alpha_deg, m, wavelength_nm=wl, grating_spacing_nm=grating_spacing_nm)
    except ValueError:
        delta = np.nan
    deviation_grating.append(delta)

# ======= 6. Plotting with separate x-axes =======
fig, axs = plt.subplots(2, 1, figsize=(12, 5))

# Prism plot
for i in range(len(wavelengths_nm) - 1):
    axs[0].axvspan(deviation_prism[i], deviation_prism[i+1], color=colors[i], linewidth=0)
axs[0].set_title("Prism Dispersion (θ₃ + θ₁ - 60°)")
axs[0].set_ylabel("Color")
axs[0].set_xlabel("Deviation Angle (degrees)")
axs[0].grid(True)

# Grating plot
for i in range(len(wavelengths_nm) - 1):
    if not np.isnan(deviation_grating[i]) and not np.isnan(deviation_grating[i+1]):
        axs[1].axvspan(deviation_grating[i], deviation_grating[i+1], color=colors[i], linewidth=0)
axs[1].set_title("Grating Dispersion (α - β)")
axs[1].set_ylabel("Color")
axs[1].set_xlabel("Deviation Angle (degrees)")
axs[1].grid(True)

plt.tight_layout()
plt.show()
```

위의 결과는 이렇게 해석이 가능하다.

1. Prism의 분산 능력은 형편없다. 가시광선의 빛이 고작해봤자 대략 2도 사이의 각도차 안에서 전부 분산된다. 

    그러나 grating의 경우 분산을 20도 넘는 각도차 내에서 분해했고, 
    
    이는 grating 사이의 간격인 grating spacing을 더 촘촘히 줄이면, 더 넓힐 수 있다. (코드를 돌려보세요.)

2. 우리는 무지개를 볼 때, 빨주노초파남보의 색 간격이 꽤 등간격으로 벌어진 것으로 본다.

    위의 prism에서 분산된 빛을 보면 납득이 된다. 어느정도 각 색의 간격이 균일하다.

    그러나 아래 grating의 결과를 보자. 빨간색이 prism에 비해 너무 길지 않은가?

    사실 우리 모두 무지개에 속고 있던 것이다. 가시광선 파장 범위를 한번 봐보자.



| 색상   | 파장 범위 (nm) | Δλ (nm) |
|--------|----------------|---------|
| 보라색 | 380 – 450      | 70      |
| 파란색 | 450 – 495      | 45      |
| 청록색 | 495 – 520      | 25      |
| 초록색 | 520 – 560      | 40      |
| 노란색 | 560 – 590      | 30      |
| 주황색 | 590 – 620      | 30      |

빨간색이 다른 색에 비해 월등하게 파장 범위가 길다.

따라서 grating이 분해한 것이 더 정확하게 나눈 것이라고 볼 수 있다.

사실 저 위에서 

"여기에서 $\alpha$나 $\beta$의 값이 충분히 작다면, radian 각 기준으로, $\Delta\theta ≈ \lambda/d$ 가 성립한다. 이것이 장점인데, radian 각에 익숙하지 않은 사람이 있을까봐, 도수법 (360도로 표현하는 방법)을 사용했다."

라고 한게 밑밥을 깐 것이다. grating의 경우 파장에 따른 deviation angle이 거의 linear하다.

이를 보여주는 graph도 한번 그려보자. 아래 graph는 입사시킨 빛의 파장에 따른 굴절된 각도 (deviation angle)을 2차 graph로 보여주는 결과이다.

```python
# 파장에 따른 deviation angle을 2차 그래프로 나타내기 위한 준비

# x축: 파장, y축: 편차각
fig, ax = plt.subplots(figsize=(10, 7))

# 프리즘 편차각
ax.plot(wavelengths_nm, deviation_prism, label="Prism (θ₃ + θ₁ − 60°)", color="blue")

# 회절격자 편차각 (NaN 제외)
valid_indices = ~np.isnan(deviation_grating)
ax.plot(wavelengths_nm[valid_indices], np.array(deviation_grating)[valid_indices],
        label="Grating (α − β)", color="red")

ax.set_title("Wavelength vs Deviation Angle")
ax.set_xlabel("Wavelength (nm)")
ax.set_ylabel("Deviation Angle (degrees)")
ax.legend()
ax.grid(True)
plt.tight_layout()
plt.show()
```

빨간색으로 표시된 결과를 보면 거의 1차 방정식에 가깝다는 것을 확인할 수 있다.

반면 파란색으로 표시된 결과는 prism의 결과인데, 미묘한 결과가 나왔다는 것을 볼 수 있다.

특히 deviation angle의 범위 (y축의 범위)를 보면 prism은 형편없이 좁다는 것을 알 수 있는데, 이것은 prism의 분해능이 굉장히 떨어진다는 것을 보여준다.

    프리즘의 경우, 특정 파장대의 빛은 분산시키지 못하고 흡수를 하기도 한다.

그렇다. 이 영역에서는 프리즘으로 빛을 분산해서 관측하는 행위 자체가 불가능하다.

그러나 회절격자의 경우, 이런 것에 꽤 자유롭다. 또한 이 영역에서조차, deviation angle은 $\lambda$와 선형 관계를 따른다.
