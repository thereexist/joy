# 음계 조율 방식별 주파수 비율 비교

| 음계 방식                        | 도 (C) | 레 (D) | 미 (E) | 파 (F) | 솔 (G) | 라 (A) | 시 (B) | 도 (C) |
|---------------------------------|--------|--------|--------|--------|--------|--------|--------|--------|
| **피타고라스 음률 (Pythagorean Tuning)** | 1      | 9/8    | 81/64  | 4/3    | 3/2    | 27/16  | 243/128 | 2      |
| **프톨레마이오스 음률 (Ptolemaic Tuning)** | 1      | 9/8    | 5/4    | 4/3    | 3/2    | 5/3    | 15/8   | 2      |
| **12평균율 (Equal Temperament)** | 1      | 1.1225 | 1.2599 | 1.3348 | 1.4983 | 1.6818 | 1.8877 | 2      |





## 1. 고대의 조율법: 피타고라스 음률 (Pythagorean Tuning)

  고대 그리스의 **피타고라스(Pythagoras)**가 개발한 조율법.

  완전 5도(3:2 비율)를 반복하여 음계를 구성.

  그러나 장3도(미-도)의 음정이 자연스럽지 않고 너무 커지는 문제점이 있음 (울림이 부자연스러움, ‘울프 톤(Wolf tone)’ 문제 발생).

피타고라스 음률은 5도를 중첩하여 만든 주파수 비율을 따름.

## 2. 르네상스 시대: 순정율 (Just Intonation) 또는 프톨레마이오스 음률

  주파수 비율이 단순한 자연수 비율을 따르는 조율법.

  예를 들어 장3도(미-도)를 5:4 비율로 조율하여 부드러운 하모니를 만듦.

  그러나 조가 바뀌면 일부 화음이 불협화음으로 들리는 문제가 발생.

## 3. 현대 음악과 평균율

  현재의 **12평균율(Equal Temperament)**은 모든 반음을 **같은 비율(100 cent)**로 나눈 방식.

  모든 조에서 동일한 화음을 만들 수 있으며, 전조가 자유롭다는 장점이 있음.

  그러나 순정율과 비교하면 특정 화음의 공명(배음 구조)이 자연스럽지 않다는 단점이 있음.

  클래식 음악에서는 경우에 따라 순정율, 중점률 등을 사용하는 경우도 있음.

그렇다면, 고대의 음과, 현대의 음은 얼마나 다르게 들릴지, 한번 python code로 확인해보도록 해보자.

```python
import numpy as np
import IPython.display as ipd
import scipy.io.wavfile as wav
import matplotlib.pyplot as plt
from IPython.display import display

# Google Colab에서 그래프 자동 출력 설정
%matplotlib inline


import numpy as np

def 주파수_12평균율(계이름, A4_freq=440.0):
    """
    현대 12평균율을 사용하여 계이름에 해당하는 주파수를 계산합니다.

    :param 계이름: 음계 이름 (예: '도', '레#', '라')
    :param A4_freq: 기준 A4의 주파수 (Hz), 기본값은 440Hz
    :return: 계산된 주파수 (Hz)
    """
    note_map = {
        '도': -9, '도#': -8, '레b': -8, '레': -7, '레#': -6, '미b': -6,
        '미': -5, '파': -4, '파#': -3, '솔b': -3, '솔': -2, '솔#': -1, '라b': -1,
        '라': 0, '라#': 1, '시b': 1, '시': 2
    }

    if 계이름 not in note_map:
        raise ValueError("Invalid note name")

    n = note_map[계이름]
    return A4_freq * (2 ** (n / 12))

def 주파수_피타고라스(계이름, A4_freq=440.0):
    """
    피타고라스 음률을 사용하여 주파수를 계산합니다.

    :param 계이름: 음계 이름 (예: '도', '레#', '라')
    :param A4_freq: 기준 A4의 주파수 (Hz)
    :return: 계산된 주파수 (Hz)
    """
    pythagorean_ratios = {
        '도': 1, '레': 9/8, '미': 81/64, '파': 4/3, '솔': 3/2, '라': 27/16, '시': 243/128,
        '도#': 256/243, '레#': 32/27, '미b': 32/27, '파#': 729/512, '솔#': 128/81, '라#': 16/9, '시b': 16/9
    }

    if 계이름 not in pythagorean_ratios:
        raise ValueError("Invalid note name")

    # C4_freq = A4_freq / (2 ** (9/12))  # C4 계산
    C4_freq = A4_freq / (27/16)  # C4 계산
    return C4_freq * pythagorean_ratios[계이름]

def 주파수_프톨레마이오스(계이름, A4_freq=440.0):
    """
    프톨레마이오스 음률을 사용하여 주파수를 계산합니다.

    :param 계이름: 음계 이름 (예: '도', '레#', '라')
    :param A4_freq: 기준 A4의 주파수 (Hz)
    :return: 계산된 주파수 (Hz)
    """
    ptolemaic_ratios = {
        '도': 1, '레': 9/8, '미': 5/4, '파': 4/3, '솔': 3/2, '라': 5/3, '시': 15/8,
        '도#': 16/15, '레#': 6/5, '미b': 6/5, '파#': 45/32, '솔#': 8/5, '라#': 9/5, '시b': 9/5
    }

    if 계이름 not in ptolemaic_ratios:
        raise ValueError("Invalid note name")

    # C4_freq = A4_freq / (2 ** (9/12))  # C4 계산
    C4_freq = A4_freq / (5/3)  # C4 계산
    return C4_freq * ptolemaic_ratios[계이름]

def generate_scale_tone(scale_function, duration=1.0, sample_rate=44100, bit_depth=32):
    """
    주어진 음계 함수에 따라 도-레-미-파-솔-라-시-도 음계를 생성하고 재생합니다.

    :param scale_function: 주어진 조율법에 따른 주파수 계산 함수
    :param scale_name: 조율법 이름
    :param duration: 각 음의 지속 시간 (초)
    :param sample_rate: 샘플링 레이트 (Hz)
    """
    scale_notes = ["도", "레", "미", "파", "솔", "라", "시", "도"]  # 1옥타브 스케일
    frequencies = [scale_function(note) for note in scale_notes[:-1]]  # 마지막 도 제외
    frequencies.append(scale_function("도") * 2)  # 마지막 도를 한 옥타브 위로 조정

    for i, frequency in enumerate(frequencies):
      print(f"{scale_notes[i]}\t {frequencies[i]}")

    t = np.linspace(0, duration * len(frequencies), int(sample_rate * duration * len(frequencies)), endpoint=False)
    wave = np.zeros_like(t)

    for i, frequency in enumerate(frequencies):
        start_sample = int(i * duration * sample_rate)
        end_sample = int((i + 1) * duration * sample_rate)
        wave[start_sample:end_sample] += 0.5 * np.sin(2 * np.pi * frequency * t[start_sample:end_sample])

    # 비트 깊이에 따라 변환
    if bit_depth == 16:
        wave_int = (wave * 32767).astype(np.int16)  # 16-bit PCM 변환
        filename = f"scale_16bit.wav"
    elif bit_depth == 24:
        wave_int = (wave * 8388607).astype(np.int32)  # 24-bit PCM 변환 (3바이트 저장)
        filename = f"scale_24bit.wav"
    elif bit_depth == 32:
        wave_int = wave.astype(np.float32)  # 32-bit float PCM 변환
        filename = f"scale_32bit.wav"
    else:
        raise ValueError("지원되지 않는 비트 깊이 (16, 24, 32만 가능)")

    wav.write(filename, sample_rate, wave_int)

    return ipd.Audio(filename)
```

# 맥놀이 효과를 이용해 각 시대별 음계가 다르다는 것을 증명.

위에서 들었듯이, 시대별 음계를 귀로 들었을 때, 민감하지 않다면 차이를 인식하지 못할 수도 있다.

이럴때 이 차이를 확인하기 위해 **맥놀이 효과**를 활용할 수 있다.

-------------------------------------------------------------------------

조율할 때 소리굽쇠 두 개의 주파수가 미묘하게 다르면 맥놀이(Beat) 현상이 발생한다.

이 현상은 두 개의 소리가 매우 비슷한 주파수를 가질 때 생기며, 서로 간섭하면서 일정한 주기의 진폭 변동이 나타난다.

예를 들어,

    하나의 소리굽쇠가 440 Hz (A음)
    다른 소리굽쇠가 442 Hz

라고 하면, 두 소리의 차이인 2 Hz의 주기로 소리가 커졌다가 작아지는 "웅웅" 같은 떨리는 느낌을 준다.

---------------------------------------------------------------

이는 수학적으로 쉽게 보일 수 있다.

두개의 음이 있다고 했을 때, 첫번째 음의 주파수가 $\omega$ 이고, 두번째 음의 주파수가 $\omega'$ 이라고 하면, 이 둘의 음을 동시에 연주했을 때 아래와 같은 wave function을 따르게 된다.

$$e^{i\omega t} + e^{i\omega' t} = e^{i\omega t}(1+e^{i(\omega' - \omega) t})$$

위의 함수는 $|\omega'-\omega|$의 주파수를 가진 sine파가 $\omega$의 주파수를 가진 sine 파로 envelope 된 형태를 나타낸다.

$\omega$와 $\omega'$ 사이의 간격이 좁을수록, "웅웅" 같은 떨리는 소리의 주파수가 증가하게 된다.

그렇기에 $\omega' - \omega$ 의 주파수로 맥놀이 현상이 발생하게 된다.

---------------------------------------------------------------


이러한 맥놀이 현상은 조율할 때 매우 유용하게 활용된다.

두 개의 소리가 완전히 같은 주파수가 되면 맥놀이가 사라지고 부드러운 단일 음으로 들리므로, 이를 통해 정확한 조율이 이루어졌는지 확인할 수 있다.

```python
def generate_tone(frequencies=[440], duration=10.0, sample_rate=44100, plot_waveform=False):
    """
    여러 주파수(frequencies)의 사인파 소리를 생성하고 Colab에서 재생합니다.

    :param frequencies: 소리의 주파수 리스트 (Hz)
    :param duration: 소리의 지속 시간 (초)
    :param sample_rate: 샘플링 레이트 (Hz)
    :param plot_waveform: 생성된 파형을 시각적으로 출력할지 여부
    """
    t = np.linspace(0, duration, int(sample_rate * duration), endpoint=False)
    wave = np.zeros_like(t)

    for frequency in frequencies:
        wave += 0.5 * np.sin(2 * np.pi * frequency * t)  # 각 주파수의 사인파를 합산

    wave /= len(frequencies)  # 볼륨을 조절하여 클리핑 방지
    wave_int = wave.astype(np.float32) # 32-bit float PCM 변환

    if plot_waveform:
        plt.figure(figsize=(20, 4))
        plt.plot(t[:2000], wave[:2000])  # 처음 1000 샘플만 표시하여 가시성 향상
        plt.title("Generated Waveform")
        plt.xlabel("Time (seconds)")
        plt.ylabel("Amplitude")
        plt.grid()
        # display(plt.gcf())  # Colab에서 명확한 출력 보장
        # plt.show()

    wav.write("output.wav", sample_rate, wave_int)
    return ipd.Audio("output.wav")
```

이런 일은 음악에서 뿐 아니라, 우리를 둘러싼 여러가지 것들에서 나타난다.

분야를 파고들면 결국 더는 증명할 수 없는 명제들이 나오고, 우리는 "그건 당연하지" 라고 비교적 쉽게 받아들인다.

이런 상황에 대해 펜로즈의 계단으로 유명한 로저 펜로즈라는 물리학자는 "인간의 이해력은 컴퓨터 계산과 분명히 다르다"는 것을 보여준다고 말한다.

나는 이런 상황이 인간이 이성적인 판단을 한다고 자부하지만, 어쩔 수 없이 intrinsic하게 가지고 있는 동물이라는 한계에 부딛히는 상황이라고 보인다.

4:5:6의 주파수비는 정말 아름다운가? 알 수 없다.

왜 우리는 청각으로 듣는 주파수는 지수적 증가를 선형 증가로 치환해서 인식하면서, 시각의 경우에는 그렇지 않는 것일까? 알 수 없다.

왜 후각은 쉽게 지쳐서 특정 향에 오래 노출되면 그 향을 강하게 느끼지 못하게 되는가? 알 수 없다.

알 수 없지만, 우리의 과학은 계속해서 이런 문제에 질문을 하고, "당연하게" 받아들여지는 명제를 더 쪼개고 쪼개는 방식으로 발전하고 있다.

이런 것이 과학의 매력이지 않을까 싶다.
